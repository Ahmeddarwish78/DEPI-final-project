# Building Fact Table

## fact_survey_response

```python
fact_survey_response = (
    df[[
        "_STATE", "SEQNO", "_METSTAT", "_URBSTAT", "IDATE",
        "_LLCPWT", "_PSU", "_STSTR", "is_complete",
        "DIABETE4", "CVDINFR4", "CVDCRHD4",
        "CVDSTRK3", "CHCKDNY2", "ADDEPEV3",
        "GENHLTH", "PHYSHLTH", "MENTHLTH",
        "HTM4", "WTKG3", "_BMI5"
    ]]
    .assign(
        date_key=lambda x: pd.to_numeric(
            pd.to_datetime(x["IDATE"], errors="coerce")
            .dt.strftime("%Y%m%d"),
            errors="coerce"
        ).astype("Int64")
    )
    .merge(
        dim_demographics[["_STATE", "SEQNO", "demographics_key"]],
        on=["_STATE", "SEQNO"],
        how="left"
    )
    .merge(
        dim_lifestyle[["_STATE", "SEQNO", "lifestyle_key"]],
        on=["_STATE", "SEQNO"],
        how="left"
    )
    .merge(
        dim_healthcare_access[
            ["_STATE", "SEQNO", "healthcare_access_key"]
        ],
        on=["_STATE", "SEQNO"],
        how="left"
    )
    .merge(
        dim_social_determinants[
            ["_STATE", "SEQNO", "social_determinants_key"]
        ],
        on=["_STATE", "SEQNO"],
        how="left"
    )
    .merge(
        dim_location[[
            "_STATE", "_METSTAT", "_URBSTAT", "location_key"
        ]],
        on=["_STATE", "_METSTAT", "_URBSTAT"],
        how="left"
    )
    .merge(
        dim_date[["date_key"]],
        on="date_key",
        how="left",
        validate="many_to_one"
    )
    .drop(columns=["_METSTAT", "_URBSTAT", "IDATE"])
    .reset_index(drop=True)
)

fact_survey_response.insert(
    0, "survey_response_key",
    fact_survey_response.index + 1
)

fact_survey_response.info()

print("\nMissing Foreign Keys:")
print(fact_survey_response[[
    "demographics_key",
    "lifestyle_key",
    "healthcare_access_key",
    "social_determinants_key",
    "location_key"
]].isnull().sum())

print("\nMissing Date Keys:",
      fact_survey_response["date_key"].isnull().sum())
```

```python
print("Duplicate Primary Keys:",
      fact_survey_response["survey_response_key"].duplicated().sum())

print("Total Rows:",
      len(fact_survey_response))

print("Unique Respondents:",
      fact_survey_response[["_STATE", "SEQNO"]].drop_duplicates().shape[0])
```

# Create the schema and tables


```python

ddl_statements = [

    # 1. Dimension: Date
    """
    IF OBJECT_ID('dbo.dim_date', 'U') IS NULL
    CREATE TABLE dbo.dim_date (
        date_key INT PRIMARY KEY,
        full_date DATE NOT NULL,
        year SMALLINT NOT NULL,
        quarter TINYINT NOT NULL,
        month SMALLINT NOT NULL,
        month_name VARCHAR(30) NOT NULL,
        day TINYINT NOT NULL,
        day_name VARCHAR(30) NOT NULL,
        week_of_year TINYINT NOT NULL
    )
    """,

    # 2. Dimension: Demographics
    """
    IF OBJECT_ID('dbo.dim_demographics', 'U') IS NULL
    CREATE TABLE dbo.dim_demographics (
        demographics_key INT PRIMARY KEY,
        _STATE VARCHAR(30),
        SEQNO BIGINT,
        SEXVAR VARCHAR(50),
        _AGE80 FLOAT,
        _RACE VARCHAR(150),
        _HISPANC VARCHAR(150),
        MARITAL VARCHAR(150),
        EDUCA VARCHAR(150),
        _EDUCAG VARCHAR(150),
        EMPLOY1 VARCHAR(150),
        INCOME3 VARCHAR(150),
        _INCOMG1 VARCHAR(150),
        RENTHOM1 VARCHAR(150),
        VETERAN3 VARCHAR(150),
        CHILDREN VARCHAR(150),
        _CHLDCNT VARCHAR(150),
        PREGNANT VARCHAR(150)
    )
    """,

    # 3. Dimension: Location
    """
    IF OBJECT_ID('dbo.dim_location', 'U') IS NULL
    CREATE TABLE dbo.dim_location (
        location_key INT PRIMARY KEY,
        _STATE VARCHAR(30),
        _METSTAT VARCHAR(100),
        _URBSTAT VARCHAR(100)
    )
    """,

    # 4. Dimension: Lifestyle
    """
    IF OBJECT_ID('dbo.dim_lifestyle', 'U') IS NULL
    CREATE TABLE dbo.dim_lifestyle (
        lifestyle_key INT PRIMARY KEY,
        _STATE VARCHAR(30),
        SEQNO BIGINT,
        SMOKE100 VARCHAR(150),
        SMOKDAY2 VARCHAR(150),
        _SMOKER3 VARCHAR(150),
        _RFSMOK3 VARCHAR(150),
        USENOW3 VARCHAR(150),
        ECIGNOW3 VARCHAR(150),
        _CURECI3 VARCHAR(150),
        LCSNUMCG FLOAT,
        DRNKANY6 VARCHAR(150),
        ALCDAY4 VARCHAR(150),
        _DRNKWK3 FLOAT,
        _RFBING6 VARCHAR(150),
        _RFDRHV9 VARCHAR(150),
        EXERANY2 VARCHAR(150),
        _TOTINDA VARCHAR(150),
        MARJSMOK VARCHAR(150)
    )
    """,

    # 5. Dimension: Healthcare Access
    """
    IF OBJECT_ID('dbo.dim_healthcare_access', 'U') IS NULL
    CREATE TABLE dbo.dim_healthcare_access (
        healthcare_access_key INT PRIMARY KEY,
        _STATE VARCHAR(30),
        SEQNO BIGINT,
        PRIMINS2 VARCHAR(150),
        PERSDOC3 VARCHAR(150),
        MEDCOST1 VARCHAR(150),
        CHECKUP1 VARCHAR(150),
        LASTDEN4 VARCHAR(150),
        RMVTETH4 VARCHAR(150),
        FLUSHOT7 VARCHAR(150),
        FLSHTMY3 VARCHAR(150),
        IMFVPLA5 VARCHAR(150),
        PNEUVAC4 VARCHAR(150),
        _HLTHPL2 VARCHAR(150),
        _HCVU654 VARCHAR(150)
    )
    """,

    # 6. Dimension: Social Determinants
    """
    IF OBJECT_ID('dbo.dim_social_determinants', 'U') IS NULL
    CREATE TABLE dbo.dim_social_determinants (
        social_determinants_key INT PRIMARY KEY,
        _STATE VARCHAR(30),
        SEQNO BIGINT,
        LSATISFY VARCHAR(150),
        EMTSUPRT VARCHAR(150),
        SDLONELY VARCHAR(150),
        SDHBILLS VARCHAR(150),
        SDHUTILS VARCHAR(150),
        SDHTRNSP VARCHAR(150),
        HOWSAFE1 VARCHAR(150),
        FOODSTMP VARCHAR(150),
        SDHFOOD1 VARCHAR(150),
        ACEDEPRS VARCHAR(150),
        CAREGIV1 VARCHAR(150),
        CIMEMLO1 VARCHAR(150),
        FIREARM5 VARCHAR(150)
    )
    """,

    # 7. Fact Table
    """
    IF OBJECT_ID('dbo.fact_survey_response', 'U') IS NULL
    CREATE TABLE dbo.fact_survey_response (
        survey_response_key BIGINT PRIMARY KEY,

        _STATE VARCHAR(30) NOT NULL,
        SEQNO BIGINT NOT NULL,

        _LLCPWT FLOAT,
        _PSU BIGINT,
        _STSTR BIGINT,
        is_complete INT,

        DIABETE4 VARCHAR(150),
        CVDINFR4 VARCHAR(150),
        CVDCRHD4 VARCHAR(150),
        CVDSTRK3 VARCHAR(150),
        CHCKDNY2 VARCHAR(150),
        ADDEPEV3 VARCHAR(150),

        GENHLTH VARCHAR(150),
        PHYSHLTH FLOAT,
        MENTHLTH FLOAT,
        HTM4 FLOAT,
        WTKG3 FLOAT,
        _BMI5 FLOAT,

        date_key INT NULL,

        demographics_key INT NOT NULL,
        lifestyle_key INT NOT NULL,
        healthcare_access_key INT NOT NULL,
        social_determinants_key INT NOT NULL,
        location_key INT NOT NULL,

        FOREIGN KEY (date_key)
            REFERENCES dbo.dim_date(date_key),

        FOREIGN KEY (demographics_key)
            REFERENCES dbo.dim_demographics(demographics_key),

        FOREIGN KEY (lifestyle_key)
            REFERENCES dbo.dim_lifestyle(lifestyle_key),

        FOREIGN KEY (healthcare_access_key)
            REFERENCES dbo.dim_healthcare_access(healthcare_access_key),

        FOREIGN KEY (social_determinants_key)
            REFERENCES dbo.dim_social_determinants(social_determinants_key),

        FOREIGN KEY (location_key)
            REFERENCES dbo.dim_location(location_key)
    )
    """
]

# Execute SQL statements
with engine.begin() as conn:
    for stmt in ddl_statements:
        conn.execute(text(stmt))

print("ALL TABLES CREATED SUCCESSFULLY!")
```

# Load Data into SQL Server Tables

```python


load_Data = [
    ("dim_date", dim_date),
    ("dim_demographics", dim_demographics),
    ("dim_location", dim_location),
    ("dim_lifestyle", dim_lifestyle),
    ("dim_healthcare_access", dim_healthcare_access),
    ("dim_social_determinants", dim_social_determinants),
    ("fact_survey_response", fact_survey_response)
]

for table_name, df in load_Data:

    with engine.connect() as conn:
        row_count = conn.execute(
            text(f"SELECT COUNT(*) FROM dbo.{table_name}")
        ).scalar()

    if row_count == len(df):
        print(f"Already Loaded: dbo.{table_name}")
        continue

    elif row_count > 0:
        print(
            f"STOP: dbo.{table_name} contains "
            f"{row_count:,} rows, expected {len(df):,}."
        )
        break

    with engine.begin() as conn:
        df.to_sql(
            table_name,
            con=conn,
            schema="dbo",
            if_exists="append",
            index=False,
            chunksize=1000
        )

    print(
        f"Loaded {len(df):,} rows --> dbo.{table_name}"
    )
```
