```python

import re
import pandas as pd

# Topics that may contribute to deeper chronic-disease analysis
keywords = [
    "blood pressure", "hypertension", "cholesterol",
    "heart attack", "coronary", "stroke", "kidney",
    "diabetes", "asthma", "copd", "lung",
    "smok", "tobacco", "e-cigarette", "alcohol",
    "exercise", "physical activity", "walking",
    "fruit", "vegetable", "nutrition", "sleep",
    "weight", "bmi", "health insurance", "medical cost",
    "checkup", "doctor", "health care", "healthcare",
    "mental health", "physical health", "depression",
    "disability", "pain", "general health",
    "income", "education", "employment", "food insecurity",
    "housing", "transportation", "social support",
    "flu shot", "pneumonia", "cancer screening"
]

# Search column labels, not the full dataset values
all_columns = pd.Index(df.columns.astype(str))
pattern = re.compile("|".join(re.escape(k) for k in keywords), re.IGNORECASE)

candidate_columns = [
    col for col in all_columns
    if pattern.search(col)
]

# Exclude columns already included in final_df
existing_columns = set(final_df.columns)
additional_candidates = [
    col for col in candidate_columns
    if col not in existing_columns
]

candidates_table = pd.DataFrame({
    "Column_Name": additional_candidates,
    "Variable_Code": [
        re.search(r"\(([^()]*)\)\s*$", col).group(1)
        if re.search(r"\(([^()]*)\)\s*$", col)
        else "Code not identified"
        for col in additional_candidates
    ]
})

print("Additional candidate columns found:", len(candidates_table))
display(candidates_table)
```

```python

import re
import pandas as pd

additional_codes = [
    "DIABAGE4", "DIABTYPE", "PREDIAB2", "DIABEDU1",
    "LCSSCNCR", "LCSLAST_", "LCSNUMC_", "_LCSAGE",
    "_LCSYSMK", "_LCSCTSN", "_LCSPSTF",
    "ALCDAY4", "AVEDRNK4",
    "FLUSHOT7", "FLSHTMY3", "IMFVPLA5", "PNEUVAC4",
    "_HLTHPL2", "_HCVU654", "_TOTINDA",
    "_LTASTH1", "_CASTHM1", "_ASTHMS1",
    "_EDUCAG", "_INCOMG1",
    "_SMOKER3", "_RFSMOK3", "_CURECI3",
    "WEIGHT2"
]


matched_columns = {}
for code in additional_codes:
    matches = [
        col for col in df.columns
        if str(col).strip().endswith(f"({code})")
        or str(col).strip() == code
    ]
    if matches:
        matched_columns[code] = matches[0]

missing_codes = [
    code for code in additional_codes
    if code not in matched_columns
]

# Add only columns that are not already in final_df
columns_to_add = [
    col for col in matched_columns.values()
    if col not in final_df.columns
]

final_df = pd.concat(
    [final_df, df[columns_to_add].copy()],
    axis=1
)

print("New columns added:", len(columns_to_add))
print("Total dataset shape:", final_df.shape)
print("Codes not found:", missing_codes)

display(final_df.head(5))
```

```python
# Count skip-routing and special-response text
special_patterns = [
    "go to section",
    "don't know",
    "not sure",
    "refused",
    "missing",
    "not asked",
    "not applicable"
]

special_rows = []

for col in data.select_dtypes(include=["object", "string"]).columns:
    values = data[col].astype("string").str.lower()

    for pattern in special_patterns:
        count = values.str.contains(
            pattern, regex=False, na=False
        ).sum()

        if count > 0:
            special_rows.append({
                "Column": col,
                "Pattern": pattern,
                "Count": int(count),
                "Percent_of_all_rows": round(
                    count / len(data) * 100, 2
                )
            })

special_values_report = pd.DataFrame(special_rows)

if not special_values_report.empty:
    display(
        special_values_report.sort_values(
            ["Column", "Count"],
            ascending=[True, False]
        )
    )
else:
    print("No matching special-response text found.")
```

```python
def find_column(code):
    matches = [
        c for c in data.columns
        if c.strip().endswith(f"({code})")
        or c.strip() == code
    ]
    return matches[0] if matches else None


variable_pairs = [
    ("ASTHMA3", "_LTASTH1"),
    ("ASTHMA3", "_CASTHM1"),
    ("ASTHNOW", "_CASTHM1"),
    ("SMOKE100", "_SMOKER3"),
    ("EXERANY2", "_TOTINDA"),
    ("PRIMINS2", "_HLTHPL2"),
    ("EDUCA", "_EDUCAG"),
    ("INCOME3", "_INCOMG1")
]

for code_a, code_b in variable_pairs:
    col_a = find_column(code_a)
    col_b = find_column(code_b)

    if col_a and col_b:
        print(f"\n{'=' * 65}")
        print(f"{code_a} vs {code_b}")

        # Compare responses only where both are available
        paired = data[[col_a, col_b]].dropna()

        print("Comparable rows:", len(paired))
        print("Agreement rate:",
              round((paired[col_a] == paired[col_b]).mean() * 100, 2),
              "%")

        display(pd.crosstab(
            paired[col_a],
            paired[col_b],
            margins=True
        ))
```

```python

import re
import pandas as pd
from pathlib import Path

# Extract the BRFSS code from the end of each descriptive column name
def extract_brfss_code(column_name):
    match = re.search(r"\(([A-Z0-9_]+)\)\s*$", str(column_name).strip())
    return match.group(1) if match else None

actual_codes = [extract_brfss_code(col) for col in data.columns]

# Check whether all columns have a recognizable BRFSS code
columns_without_codes = [
    col for col, code in zip(data.columns, actual_codes)
    if code is None
]

# Check duplicate codes
valid_codes = [code for code in actual_codes if code is not None]
duplicate_codes = pd.Series(valid_codes)
duplicate_codes = duplicate_codes[duplicate_codes.duplicated()].unique().tolist()

# Expected codes from the previously documented variable list
expected_codes = set(expected_variables)
actual_code_set = set(valid_codes)

missing_codes = sorted(expected_codes - actual_code_set)
unexpected_codes = sorted(actual_code_set - expected_codes)

# Create a report for every actual column
verification_report = pd.DataFrame({
    "Original_Column_Name": list(data.columns),
    "Extracted_BRFSS_Code": actual_codes,
    "Code_Recognized": [code is not None for code in actual_codes],
    "Code_In_Expected_List": [
        code in expected_codes if code is not None else False
        for code in actual_codes
    ],
    "Data_Type": [str(data[col].dtype) for col in data.columns],
    "Missing_Values": [int(data[col].isna().sum()) for col in data.columns],
    "Unique_Values": [int(data[col].nunique(dropna=True)) for col in data.columns]
})

# Save the report
output_dir = Path.cwd() / "BRFSS_Validation_Reports"
output_dir.mkdir(parents=True, exist_ok=True)

verification_report.to_csv(
    output_dir / "variable_code_verification.csv",
    index=False,
    encoding="utf-8-sig"
)

# Print results
print("=" * 60)
print("BRFSS VARIABLE CODE VERIFICATION")
print("=" * 60)
print(f"Dataset shape: {data.shape}")
print(f"Total columns: {len(data.columns)}")
print(f"Recognized BRFSS codes: {sum(code is not None for code in actual_codes)}")
print(f"Columns without recognized codes: {len(columns_without_codes)}")
print(f"Unique extracted codes: {len(actual_code_set)}")
print(f"Duplicate extracted codes: {duplicate_codes}")
print(f"Expected codes missing: {len(missing_codes)}")
print(f"Unexpected codes: {len(unexpected_codes)}")

if missing_codes:
    print("\nMissing expected codes:")
    print(missing_codes)

if unexpected_codes:
    print("\nUnexpected extracted codes:")
    print(unexpected_codes)

if columns_without_codes:
    print("\nColumns without recognizable codes:")
    print(columns_without_codes)

print("\nFirst 10 mappings:")
print(verification_report[
    ["Original_Column_Name", "Extracted_BRFSS_Code"]
].head(10).to_string(index=False))

print("\nReport saved to:")
print((output_dir / "variable_code_verification.csv").resolve())
```
